```python
shipment_df = spark.table(SHIPMENT)
```

```python
shipment_delay_df = (
    shipment_df
    .withColumn(
        "delay_flag",
        F.when(
            F.col("delivery_delay_days") > 0,
            "DELAYED"
        )
        .otherwise(
            "ON_TIME"
        )
    )
)
```

```python
shipment_delay_df = (
    shipment_df
    .withColumn(
        "delay_flag",
        F.when(
            F.col("delivery_delay_days") > 0,
            "DELAYED"
        )
        .otherwise(
            "ON_TIME"
        )
    )
)
```

```python
shipment_delay_df = (
    shipment_delay_df.select(
        "shipment_id",
        "source_depot_id",
        "destination_site_id",
        "drug_type",
        "current_status",
        "expected_delivery_date",
        "actual_delivery_date",
        "delivery_delay_days",
        "delay_flag"
    )
)
```
