```python
from pyspark.sql import functions as F
```

```python
CATALOG = "clinical_trial_supply_chain"

BRONZE = f"{CATALOG}.bronze"
SILVER = f"{CATALOG}.silver"

SOURCE_TABLE = f"{BRONZE}.bronze_dispensing"
TARGET_TABLE = f"{SILVER}.fact_dispensing"
```

```python
dispensing_df = (
    spark.table(SOURCE_TABLE)
    .select(
        "dispense_id",
        "patient_id",
        "site_id",
        "study_id",
        "drug_type",
        "kit_id",
        "visit_number",
        "dispense_date",
        "quantity_dispensed"
    )
    .dropDuplicates()
)
```

```python
(
    dispensing_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(TARGET_TABLE)
)
```

```python
print(
    f"fact_dispensing records : "
    f"{spark.table(TARGET_TABLE).count()}"
)

display(
    spark.table(TARGET_TABLE)
)
```
