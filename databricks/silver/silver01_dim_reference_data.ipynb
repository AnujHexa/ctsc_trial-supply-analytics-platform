```python
from pyspark.sql import functions as F
```

```python
CATALOG = "clinical_trial_supply_chain"

BRONZE = f"{CATALOG}.bronze"
SILVER = f"{CATALOG}.silver"
```

```python
country_df = (
    spark.table(f"{BRONZE}.bronze_country_master")
    .select(
        "country_code",
        "country_name",
        "region"
    )
    .dropDuplicates()
)

(
    country_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(f"{SILVER}.dim_country")
)

print(f"dim_country loaded : {country_df.count()} records")
```

```python
depot_df = (
    spark.table(f"{BRONZE}.bronze_depot_master")
    .select(
        "depot_id",
        "depot_name",
        "country_code",
        "region"
    )
    .dropDuplicates()
)

(
    depot_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(f"{SILVER}.dim_depot")
)

print(f"dim_depot loaded : {depot_df.count()} records")
```

```python
drug_df = (
    spark.table(f"{BRONZE}.bronze_drug_master")
    .select(
        "drug_type",
        "drug_name",
        "shelf_life_days",
        "reorder_threshold",
        "unit_of_measure"
    )
    .dropDuplicates()
)

(
    drug_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(f"{SILVER}.dim_drug")
)

print(f"dim_drug loaded : {drug_df.count()} records")
```

```python
print("dim_country")
display(
    spark.table(f"{SILVER}.dim_country")
)

print("dim_depot")
display(
    spark.table(f"{SILVER}.dim_depot")
)

print("dim_drug")
display(
    spark.table(f"{SILVER}.dim_drug")
)
```
