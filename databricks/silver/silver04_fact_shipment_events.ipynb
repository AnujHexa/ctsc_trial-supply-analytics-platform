```python
from pyspark.sql import functions as F
```

```python
CATALOG = "clinical_trial_supply_chain"

BRONZE = f"{CATALOG}.bronze"
SILVER = f"{CATALOG}.silver"

SOURCE_TABLE = f"{BRONZE}.bronze_shipment_events"
TARGET_TABLE = f"{SILVER}.fact_shipment_events"
```

```python
shipment_df = (
    spark.table(SOURCE_TABLE)
    .select(
        "shipment_id",
        "shipment_event_timestamp",
        "source_depot_id",
        "destination_site_id",
        "study_id",
        "drug_type",
        "qty_shipped",
        "shipment_status",
        "expected_delivery_date",
        "actual_delivery_date"
    )
    .dropDuplicates()
)
```

```python
(
    shipment_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(TARGET_TABLE)
)
```

```python
print(
    f"fact_shipment_events records : "
    f"{spark.table(TARGET_TABLE).count()}"
)

display(
    spark.table(TARGET_TABLE)
    .orderBy(
        "shipment_id",
        "shipment_event_timestamp"
    )
)
```
